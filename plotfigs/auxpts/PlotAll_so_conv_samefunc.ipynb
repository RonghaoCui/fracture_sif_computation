```python
# Only for plot SIF results for all grid sizes
from __future__ import annotations
from dataclasses import dataclass

import numpy as np
import math
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.ticker import ScalarFormatter
from matplotlib.ticker import LogLocator, LogFormatterMathtext

# MPSA
mpsa_dofs = np.array([
    16.32151574, 31.82437906, 62.97294016, 125.48940643, 250.10555251, 499.25663341
])

err_ave_mpsa = np.array([
    0.01560495, 0.00494485, 0.00304355, 0.0023395, 0.00123267, 0.00078003
]) 

# From Fitting parameters: C=1.00, alpha=0.25
mpsa_dofs_2 = np.array([
    16.32151574, 31.82437906, 62.97294016, 125.48940643, 250.10555251, 499.25663341
])

err_ave_mpsa_2 = np.array([
    0.05203555, 0.01516472, 0.01140695, 0.00796419, 0.00509855, 0.0034752
]) 

# QFEM-QPE
# From Fitting parameters: C=1.0671, alpha=0.3760
femqpe_dofs = np.array([
     33.48983621, 64.52259576, 126.83203562, 251.86973768, 501.1059267 
])

err_ave_femqpe = np.array([
    0.0094868, 0.00209868, 0.00053325, 0.00040623, 0.00024107
]) 

# From Fitting parameters: C=1.00, alpha=0.25
femqpe_dofs_2 = np.array([
    33.48983621, 64.52259576, 126.83203562, 251.86973768, 501.1059267 
])

err_ave_femqpe_2 = np.array([
    0.02238185, 0.01227281, 0.00781963, 0.00620559, 0.00427149
]) 


# QFEM no qpe
fem_dofs = np.array([
    33.48983621, 64.52259576, 126.83203562, 251.86973768, 501.1059267
])

err_ave_fem = np.array([
    0.00857427, 0.00148248, 0.00120427, 0.00049011, 0.00029562
]) 
# From Fitting parameters: C=1.00, alpha=0.25
fem_dofs_2 = np.array([
    33.48983621, 64.52259576, 126.83203562, 251.86973768, 501.1059267
])

err_ave_fem_2 = np.array([
    0.03220765, 0.01153264, 0.00432891, 0.00294177, 0.00165023
]) 

# Reference line
num_dofs_line = np.array([10, 30, 50, 70, 90, 100, 300, 500, 1000])
initial_value_line = 1e-3
first_order_line = initial_value_line * (num_dofs_line[0] / num_dofs_line)
half_order_line = initial_value_line  * (num_dofs_line[0] / num_dofs_line) ** 0.5

# Plot SIF KI
mpl.rcParams.update({
    "font.family": "serif",     
    "font.size": 8,             
    "axes.labelsize": 8,       
    "axes.titlesize": 8,        
    "legend.fontsize": 8,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "lines.linewidth": 1.0,      
    "lines.markersize": 3.0,       
    "mathtext.fontset": "dejavusans",  
})
formatter = ScalarFormatter(useMathText=True)
formatter.set_powerlimits((0, 0))   # Scientific form

fig, ax = plt.subplots(figsize=(5.5, 3.0))   # inches
# MPSA
ax.loglog(
    mpsa_dofs,
    err_ave_mpsa,
    'x-',
    color='tab:blue',
    linewidth=1.5, 
    markersize=4.0,
    label="MPSA (C=0.73, alpha=0.30)"
)

ax.loglog(
    mpsa_dofs_2,
    err_ave_mpsa_2,
    'x--',
    color='tab:blue',
    linewidth=1.5, 
    markersize=4.0,
    label="MPSA (C=1.00, alpha=0.25)"
)

# QFEM-QPE
ax.loglog(
    femqpe_dofs,
    err_ave_femqpe,
    'x-',
    color='tab:orange',
    linewidth=1.5, 
    markersize=4.0,
    label="QFEM-QPE (C=1.07, alpha=0.38)"
)

ax.loglog(
    femqpe_dofs_2,
    err_ave_femqpe_2,
    'x--',
    color='tab:orange',
    linewidth=1.5, 
    markersize=4.0,
    label="QFEM-QPE (C=1.00, alpha=0.25)"
)

# QFEM
ax.loglog(
    fem_dofs,
    err_ave_fem,
    'x-',
    color='tab:pink',
    linewidth=1.5, 
    markersize=4.0,
    label="QFEM (C=0.90, alpha=0.25)"
)

ax.loglog(
    fem_dofs_2,
    err_ave_fem_2,
    'x--',
    color='tab:pink',
    linewidth=1.5, 
    markersize=4.0,
    label="QFEM (C=1.00, alpha=0.25)"
)

ax.loglog(
    num_dofs_line, first_order_line,
    '--', color='k',
    linewidth=2.0,
    label="Reference slope=-1"
)

ax.xaxis.set_major_formatter(formatter)
ax.yaxis.set_major_formatter(formatter)
ax.set_xlim(10, 700)
ax.set_ylim(0.00005, 1.0)
ax.xaxis.set_major_locator(LogLocator(base=10))
ax.xaxis.set_major_formatter(LogFormatterMathtext())
ax.xaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10) * 0.1))

ax.yaxis.set_major_locator(LogLocator(base=10))
ax.yaxis.set_major_formatter(LogFormatterMathtext())
ax.yaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10) * 0.1))

ax.set_xlabel(r"$\sqrt{\mathrm{DOF}}$")
ax.set_ylabel("Relative error")

# ax.legend(loc="lower left")
ax.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5),  
    frameon=False
)
ax.grid(True, which="both", linestyle="--", alpha=0.5)
plt.tight_layout()
# plt.savefig("KI_convergence_so.tiff", format="tiff", dpi=300, bbox_inches="tight", pil_kwargs={"compression": "tiff_lzw"})
# plt.savefig("sneddon_singularity.eps", format="eps", dpi=300, bbox_inches="tight")
plt.show()
```
