```python
# Only for plot SIF results for all grid sizes
from __future__ import annotations
from dataclasses import dataclass

import numpy as np
import math
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.ticker import ScalarFormatter
from matplotlib.ticker import LogLocator, LogFormatterMathtext

# MPSA
mpsa_dofs = np.array([
    16.00, 31.43246729, 64.49806199, 122.01639234, 244.03688246, 489.24227127
])

err_ave_mpsa = np.array([
    0.00689743, 0.00275244, 0.00185918, 0.00058247, 0.00080624, 0.00051805
]) 

err_std_mpsa = np.array([
    1.26085837e-03, 4.37837875e-04, 1.80149185e-04, 1.21126523e-04, 4.59791441e-05, 2.36704094e-05
]) 

# TPSA
tpsa_dofs = np.array([
    16.00, 31.43246729, 61.28621378, 122.01639234, 244.03688246, 487.59614437
])

err_ave_tpsa = np.array([
    0.06518692, 0.02962932, 0.02693867, 0.01599449, 0.00902199, 0.00227075
]) 

err_std_tpsa = np.array([
    0.00605779, 0.00396379, 0.00236685, 0.00070604, 0.00122892, 0.00141808
]) 

# QFEM-QPE
femqpe_dofs = np.array([
    32.86335345, 63.7495098, 123.48279232, 244.94897428, 488.99079746
])

err_ave_femqpe = np.array([
    9.06732843e-03, 2.18018448e-03, 2.22204085e-03, 3.96319889e-04, 8.45424809e-05
]) 

err_std_femqpe = np.array([
    2.38213839e-04, 1.11012759e-04, 7.42690946e-05, 8.87464164e-06, 7.92811131e-06
]) 

# QFEM no qpe
fem_dofs = np.array([
    32.86335345, 63.7495098, 122.99593489, 244.96530367, 488.90080794
])

err_ave_fem = np.array([
    9.65656694e-03, 1.63391590e-03, 1.53823686e-03, 5.06083612e-04, 9.68362041e-05
]) 

err_std_fem = np.array([
    2.51801522e-04, 1.15900947e-04, 5.79159247e-05, 1.90461186e-05, 9.70667817e-06
]) 

# Reference line
num_dofs_line = np.array([10, 30, 50, 70, 90, 100, 300, 500, 1000])
initial_value_line = 3e-4
first_order_line = initial_value_line * (num_dofs_line[0] / num_dofs_line)
half_order_line = initial_value_line  * (num_dofs_line[0] / num_dofs_line) ** 0.5

# Plot SIF KI
mpl.rcParams.update({
    "font.family": "serif",     
    "font.size": 8,             
    "axes.labelsize": 8,       
    "axes.titlesize": 8,        
    "legend.fontsize": 8,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "lines.linewidth": 1.0,      
    "lines.markersize": 3.0,       
    "mathtext.fontset": "dejavusans",  
})
formatter = ScalarFormatter(useMathText=True)
formatter.set_powerlimits((0, 0))   # Scientific form

fig, ax = plt.subplots(figsize=(5.0, 3.0))   # inches
# MPSA
ax.loglog(
    mpsa_dofs,
    err_ave_mpsa,
    'x-',
    color='tab:blue',
    linewidth=2.0, 
    markersize=4.0,
    label="MPSA (ave)"
)

ax.fill_between(
    mpsa_dofs,
    err_ave_mpsa - err_std_mpsa,
    err_ave_mpsa + err_std_mpsa,
    color='tab:blue',
    alpha=0.35,
    label="MPSA: ±1 std"
)

# TPSA
ax.loglog(
    tpsa_dofs,
    err_ave_tpsa,
    'x-',
    color='tab:green',
    linewidth=2.0, 
    markersize=4.0,
    label="TPSA (ave)"
)

ax.fill_between(
    tpsa_dofs,
    err_ave_tpsa - err_std_tpsa,
    err_ave_tpsa + err_std_tpsa,
    color='tab:green',
    alpha=0.35,
    label="TPSA: ±1 std"
)

# QFEM-QPE
ax.loglog(
    femqpe_dofs,
    err_ave_femqpe,
    'x-',
    color='tab:orange',
    linewidth=2.0, 
    markersize=4.0,
    label="QFEM-QPE (ave)"
)

ax.fill_between(
    femqpe_dofs,
    err_ave_femqpe - err_std_femqpe,
    err_ave_femqpe + err_std_femqpe,
    color='tab:orange',
    alpha=0.35,
    label="QFEM-QPE: ±1 std"
)

# QFEM
ax.loglog(
    fem_dofs,
    err_ave_fem,
    'x-',
    color='tab:pink',
    linewidth=2.0, 
    markersize=4.0,
    label="QFEM (ave)"
)

ax.fill_between(
    fem_dofs,
    err_ave_fem - err_std_fem,
    err_ave_fem + err_std_fem,
    color='tab:pink',
    alpha=0.35,
    label="QFEM: ±1 std"
)

ax.loglog(
    num_dofs_line, first_order_line,
    '--', color='k',
    linewidth=2.0,
    label="Reference slope=-1"
)

ax.xaxis.set_major_formatter(formatter)
ax.yaxis.set_major_formatter(formatter)
ax.set_xlim(10, 700)
ax.set_ylim(0.00002, 1.0)
ax.xaxis.set_major_locator(LogLocator(base=10))
ax.xaxis.set_major_formatter(LogFormatterMathtext())
ax.xaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10) * 0.1))

ax.yaxis.set_major_locator(LogLocator(base=10))
ax.yaxis.set_major_formatter(LogFormatterMathtext())
ax.yaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10) * 0.1))

ax.set_xlabel(r"$\sqrt{\mathrm{DOF}}$")
ax.set_ylabel("Relative error")

# ax.legend(loc="lower left")
ax.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5),  
    frameon=False
)
ax.grid(True, which="both", linestyle="--", alpha=0.5)
plt.tight_layout()
# plt.savefig("KI_convergence_so.tiff", format="tiff", dpi=300, bbox_inches="tight", pil_kwargs={"compression": "tiff_lzw"})
# plt.savefig("sneddon_singularity.eps", format="eps", dpi=300, bbox_inches="tight")
plt.show()
```
